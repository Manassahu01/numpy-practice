# 🔍 Sorting, Searching & Filtering in NumPy

```
import numpy as np
```

```
arr=np.array([40,10,30,20,50,30])
print(arr)
```

## np.sort()

```
print(np.sort(arr))
```

## np.argsort()

```
print(np.argsort(arr))
```

## np.where()

```
print(np.where(arr==30))
print(np.where(arr>25))
```

## np.searchsorted()

```
s=np.sort(arr)
print(s)
print(np.searchsorted(s,35))
```

## Boolean Filtering

```
print(arr[arr>25])
print(arr[arr%2==0])
```

## Conditional Filtering

```
marks=np.array([45,67,82,91,55,39])
print(marks[marks>=40])
print(marks[marks>80])
```

## Duplicate Values

```
u,c=np.unique(arr,return_counts=True)
print(u[c>1])
```

## Summary
- np.sort()
- np.argsort()
- np.where()
- np.searchsorted()
- Boolean Filtering
- Conditional Filtering
- Duplicate Values
