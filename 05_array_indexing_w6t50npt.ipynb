# 🔍 Array Indexing in NumPy

Learn how to access elements in NumPy arrays using indexing.

```
import numpy as np
```

## Create Arrays

```
arr1 = np.array([10, 20, 30, 40, 50])

arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])

arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
```

## 1. Positive Indexing (1D Array)

```
print(arr1[0])
print(arr1[2])
print(arr1[4])
```

## 2. Negative Indexing (1D Array)

```
print(arr1[-1])
print(arr1[-2])
print(arr1[-3])
```

## 3. Indexing in 2D Array

```
print(arr2[0, 0])
print(arr2[0, 2])
print(arr2[1, 1])
print(arr2[1, 2])
```

## 4. Indexing in 3D Array

```
print(arr3[0, 0, 0])
print(arr3[0, 1, 1])
print(arr3[1, 0, 1])
print(arr3[1, 1, 0])
```

## 5. Access Complete Row

```
print(arr2[0])
print(arr2[1])
```

## 6. Access Complete Column

```
print(arr2[:, 0])
print(arr2[:, 1])
print(arr2[:, 2])
```

## 7. Modify Elements

```
print("Before:")
print(arr1)

arr1[2] = 100

print("After:")
print(arr1)
```

## ✅ Summary

- 3D Array Indexing
- Access Rows
- Access Columns
