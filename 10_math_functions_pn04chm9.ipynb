# 🧮 Mathematical Functions in NumPy

Learn the most commonly used mathematical functions in NumPy.

```
import numpy as np
```

## Create an Array

```
arr = np.array([4, 9, 16, 25, 36])
print(arr)
```

## 1. np.sum()

```
print(np.sum(arr))
```

## 2. np.min() and np.max()

```
print(np.min(arr))
print(np.max(arr))
```

## 3. np.sqrt()

```
print(np.sqrt(arr))
```

## 4. np.power()

```
print(np.power(arr, 2))
print(np.power(arr, 3))
```

## 5. np.abs()

```
numbers = np.array([-10, -5, 0, 5, 10])
print(np.abs(numbers))
```

## 6. np.round()

```
values = np.array([2.345, 4.567, 8.912])
print(np.round(values, 2))
```

## 7. np.ceil() and np.floor()

```
values = np.array([2.3, 4.8, 6.1])

print(np.ceil(values))
print(np.floor(values))
```
