# 📝 NumPy Practice Programs

Practice the concepts learned in previous NumPy notebooks.

```
import numpy as np
```

## Program 1: Create a 1D Array

```
arr = np.array([10,20,30,40,50])
print(arr)
```

## Program 2: Create a 3×3 Matrix of Zeros

```
print(np.zeros((3,3), dtype=int))
```

## Program 3: Create Numbers from 1 to 20

```
print(np.arange(1,21))
```

## Program 4: Reshape Array

```
arr=np.arange(1,13)
print(arr.reshape(3,4))
```

## Program 5: Find Mean, Median and Standard Deviation

```
arr=np.array([12,15,18,21,24])
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
```

## Program 6: Find Maximum and Minimum

```
arr=np.array([4,9,2,7,1])
print(np.max(arr))
print(np.min(arr))
```

## Program 7: Sort an Array

```
arr=np.array([5,2,9,1,7])
print(np.sort(arr))
```

## Program 8: Filter Even Numbers

```
arr=np.arange(1,21)
print(arr[arr%2==0])
```

## Program 9: Random Integers

```
print(np.random.randint(1,101,size=10))
```

## Program 10: Matrix Addition

```
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a+b)
```

## Program 11: Matrix Multiplication

```
print(a@b)
```

## Program 12: Transpose

```
print(a.T)
```

## Program 13: Square Root

```
arr=np.array([1,4,9,16,25])
print(np.sqrt(arr))
```

## Program 14: Cumulative Sum

```
arr=np.array([10,20,30,40])
print(np.cumsum(arr))
```

## Program 15: Boolean Filtering

```
arr=np.array([15,42,67,89,23,91])
print(arr[arr>50])
```
