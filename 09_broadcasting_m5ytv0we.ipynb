# 📡 Broadcasting in NumPy

Learn how NumPy automatically broadcasts arrays.

```
import numpy as np
```

## 1. Scalar Broadcasting

```
arr=np.array([10,20,30,40])
print(arr+5)
print(arr*2)
```

## 2. 1D + 2D Broadcasting

```
a=np.array([[1,2,3],[4,5,6]])
b=np.array([10,20,30])
print(a+b)
```

## 3. Column Vector Broadcasting

```
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[10],[20]])
print(a+b)
```

## 4. Broadcasting Rules

```
print("1. Compare dimensions from right to left")
print("2. Dimensions must be equal or 1")
print("3. Otherwise ValueError")
```

## 5. Shape Mismatch Example

```
x=np.array([1,2,3])
y=np.array([1,2])
try:
    print(x+y)
except ValueError as e:
    print("Error:",e)
```

## 6. Practical Example

```
marks=np.array([45,55,65,75])
print(marks+5)
```
