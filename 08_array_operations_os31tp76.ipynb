# ➕ Array Operations in NumPy

Learn how to perform arithmetic and comparison operations on NumPy arrays.

```
import numpy as np
```

## Create Arrays

```
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([1, 2, 3, 4])

print("arr1 =", arr1)
print("arr2 =", arr2)
```

## 1. Addition

```
print(arr1 + arr2)
print(np.add(arr1, arr2))
```

## 2. Subtraction

```
print(arr1 - arr2)
print(np.subtract(arr1, arr2))
```

## 3. Multiplication

```
print(arr1 * arr2)
print(np.multiply(arr1, arr2))
```

## 4. Division

```
print(arr1 / arr2)
print(np.divide(arr1, arr2))
```

## 5. Floor Division, Modulus and Power

```
print(arr1 // arr2)
print(arr1 % arr2)
print(arr2 ** 2)
```

## 6. Array with Scalar

```
print(arr1 + 5)
print(arr1 * 10)
```

## 7. Comparison Operators

```
print(arr1 > 20)
print(arr1 == 20)
print(arr1 != 20)
```

## 8. Logical Operations

```
print(np.logical_and(arr1 > 10, arr1 < 40))
print(np.logical_or(arr1 < 20, arr1 > 30))
print(np.logical_not(arr1 > 20))
```
