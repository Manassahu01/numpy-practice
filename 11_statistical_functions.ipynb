# 📊 Statistical Functions in NumPy

Learn the most commonly used statistical functions in NumPy.

```
import numpy as np
```

## Create an Array

```
arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

## 1. Mean

```
print("Mean :", np.mean(arr))
```

## 2. Median

```
print("Median :", np.median(arr))
```

## 3. Standard Deviation

```
print("Standard Deviation :", np.std(arr))
```

## 4. Variance

```
print("Variance :", np.var(arr))
```

## 5. Percentile

```
print("25th Percentile :", np.percentile(arr, 25))
print("50th Percentile :", np.percentile(arr, 50))
print("75th Percentile :", np.percentile(arr, 75))
```

## 6. Index of Maximum Value

```
print("Index of Maximum Value :", np.argmax(arr))
```

## 7. Index of Minimum Value

```
print("Index of Minimum Value :", np.argmin(arr))
```

## 8. Cumulative Sum

```
print(np.cumsum(arr))
```

## 9. Statistical Functions on 2D Arrays

```
matrix = np.array([[10,20,30],[40,50,60]])
print("Mean:", np.mean(matrix))
print("Row-wise Mean:", np.mean(matrix, axis=1))
print("Column-wise Mean:", np.mean(matrix, axis=0))
```

## 10. Practice Example

```
marks = np.array([78,85,92,67,88,95])
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Std Dev:", np.std(marks))
```
