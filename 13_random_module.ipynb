# 🎲 Random Module in NumPy

```
import numpy as np
```

## np.random.rand()

```
print(np.random.rand())
print(np.random.rand(3))
print(np.random.rand(2,3))
```

## np.random.randn()

```
print(np.random.randn(5))
```

## np.random.randint()

```
print(np.random.randint(1,10))
print(np.random.randint(1,100,size=5))
print(np.random.randint(1,20,size=(2,3)))
```

## np.random.choice()

```
items=['Apple','Banana','Mango','Orange']
print(np.random.choice(items))
print(np.random.choice(items,size=3))
```

## np.random.shuffle()

```
arr=np.array([1,2,3,4,5])
print(arr)
np.random.shuffle(arr)
print(arr)
```

## np.random.seed()

```
np.random.seed(42)
print(np.random.randint(1,100,5))
np.random.seed(42)
print(np.random.randint(1,100,5))
```

## Practice

```
marks=np.random.randint(35,101,10)
print(marks)
print(marks.mean())
```
