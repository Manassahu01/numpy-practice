# 📊 Array Attributes in NumPy

Learn the important attributes of NumPy arrays.

```
import numpy as np
```

## Create an Array

```
arr=np.array([[10,20,30],[40,50,60]])
print(arr)
```

## 1. ndim

```
print(arr.ndim)
```

## 2. shape

```
print(arr.shape)
```

## 3. size

```
print(arr.size)
```

## 4. dtype

```
print(arr.dtype)
```

## 5. itemsize

```
print(arr.itemsize)
```

## 6. nbytes

```
print(arr.nbytes)
```

## 7. astype()

```
b=arr.astype(float)
print(b)
print(b.dtype)
```

## Summary
- ndim
- shape
- size
- dtype
- itemsize
- nbytes
- astype()
