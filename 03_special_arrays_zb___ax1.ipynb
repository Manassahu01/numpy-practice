# ⭐ Special Arrays in NumPy

This notebook covers commonly used NumPy functions for creating special arrays.

```
import numpy as np
```

## 1. np.zeros()

```
arr = np.zeros((3, 4), dtype=int)
print(arr)
```

## 2. np.ones()

```
arr = np.ones((2, 5), dtype=float)
print(arr)
```

## 3. np.empty()

```
arr = np.empty((2, 3))
print(arr)
# Values are uninitialized
```

## 4. np.full()

```
arr = np.full((3, 3), 7)
print(arr)
```

## 5. np.eye()

```
arr = np.eye(4, dtype=int)
print(arr)
```

## 6. np.identity()

```
arr = np.identity(4, dtype=int)
print(arr)
```

## 7. np.arange()

```
print(np.arange(10))
print(np.arange(2, 20, 3))
```

## 8. np.linspace()

```
print(np.linspace(0, 1, 5))
print(np.linspace(10, 100, 10))
```
